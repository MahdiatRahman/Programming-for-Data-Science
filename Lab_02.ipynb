# Problem 1 – Course Marks

```python
marks = [75, 82, 68, 91, 45]

print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))

passed = 0
for mark in marks:
    if mark >= 50:
        passed += 1

print("Passed courses:", passed)
```

# Problem 3 – Word Analyzer

```python
text = input("Enter a word or sentence: ")

vowels = 0
spaces = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    if ch == " ":
        spaces += 1

print("Characters:", len(text))
print("Vowels:", vowels)
print("Spaces:", spaces)
```

# Problem 4 – Grade Calculator

```python
def calculate_grade(marks):
    avg = sum(marks) / len(marks)

    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    else:
        return "F"

marks = [80, 75, 90, 65, 70]
print("Grade:", calculate_grade(marks))
```

# Problem 5 – Shopping Cart Processor

```python
products = [
    {"name": "Bag", "price": 1200, "quantity": 1},
    {"name": "Shoes", "price": 1000, "quantity": 1},
    {"name": "Pen", "price": 20, "quantity": 2},
    {"name": "Book", "price": 200, "quantity": 1}
]

total = 0

for p in products:
    total += p["price"] * p["quantity"]

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    return total

print("Original total:", total)
print("Final total:", apply_discount(total))
```

# Problem 6 – Countdown

# Problem 7 – Student Record Analyzer

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total = 0
passed = 0

for s in students:
    print(s["name"], s["marks"])
    total += s["marks"]

    if s["marks"] >= 50:
        passed += 1

top = max(students, key=lambda s: s["marks"])

print("Passed:", passed)
print("Class average:", total / len(students))
print("Highest scorer:", top["name"])
```

# Problem 8 – Product Inventory Summary

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0

for p in products:
    line_total = p["price"] * p["quantity"]
    print(p["name"], ":", line_total)
    total += line_total

top = max(products, key=lambda p: p["price"] * p["quantity"])

print("Total inventory value:", total)
print("Highest line total:", top["name"])
```
